```python
'''Problem Statement:
The aim of this project is to create a simple program to track daily expenses.
The program takes expense values from the user, stores them in a NumPy array,
and calculates total expense, average expense, and highest expense.
It also displays a spending summary and gives a warning if average expense is high.'''
```

```python
'''Logic Explanation:
1. The user enters the number of days.
2. Daily expenses are taken using a loop and stored in a list.
3. The list is converted into a NumPy array.
4. NumPy functions are used to calculate total, average, and highest expense.
5. A loop is used to find the day with the highest expense.
6. If the average expense exceeds a limit, a warning is displayed.'''
```

```python
import numpy as np

def analyze_expenses(expense_array, days):
    total = expense_array.sum()
    average = expense_array.mean()
    highest = expense_array.max()

    # Find highest expense day using loop
    max_index = 0
    for i in range(days):
        if expense_array[i] > expense_array[max_index]:
            max_index = i

    print("\n--- Daily Expense Report ---")
    print("Expenses:", list(expense_array))
    print("Total Expense:", total)
    print("Average Expense:", round(average, 2))
    print("Highest Expense:", highest)
    print("Highest Expense Day: Day", max_index + 1)

    # Optional warning
    limit = 500
    if average > limit:
        print("Warning: Your average expense is too high!")

# Input
days = int(input("Enter number of days: "))
expense_list = []

for i in range(days):
    amount = float(input("Enter expense for day: "))
    expense_list.append(amount)

# Convert list to NumPy array
expense_array = np.array(expense_list)

# Function call
analyze_expenses(expense_array, days)
```
