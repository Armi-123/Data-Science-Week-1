```python
'''Problem Statement:
The purpose of this project is to analyze marks of multiple students using Python and NumPy.
The program takes marks as input, stores them in a NumPy array, and calculates total marks,
average marks, highest and lowest marks. Based on the average, a grade is assigned using
conditional statements.'''
```

```python
'''Logic Explanation:
1. The user enters the number of students.
2. Marks are taken using a loop and stored in a list.
3. The list is converted into a NumPy array.
4. NumPy functions are used to find total, average, highest, and lowest marks.
5. Conditional statements are used to assign grade based on average marks.
6. The final report is displayed to the user.'''
```

```python
import numpy as np

def calculate_grade(avg):
    if avg > 85:
        return "A"
    elif avg >= 70:
        return "B"
    elif avg >= 50:
        return "C"
    else:
        return "Fail"

# Input number of students
n = int(input("Enter number of students: "))

marks_list = []

# Taking marks input
for i in range(n):
    mark = float(input("Enter marks of student: "))
    marks_list.append(mark)

# Convert list to NumPy array
marks_array = np.array(marks_list)

# Calculations using NumPy
total = marks_array.sum()
average = marks_array.mean()
highest = marks_array.max()
lowest = marks_array.min()

# Grade calculation
grade = calculate_grade(average)

# Output
print("\n--- Student Marks Report ---")
print("Marks:", list(marks_array))
print("Total Marks:", total)
print("Average Marks:", round(average, 2))
print("Highest Marks:", highest)
print("Lowest Marks:", lowest)
print("Grade:", grade)
```
