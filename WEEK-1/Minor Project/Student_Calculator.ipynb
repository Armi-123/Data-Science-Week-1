```python
'''Project Title:Console-Based Smart Calculator using Python and NumPy'''
```

```python
'''Problem Statement:
The goal of this project is to build a menu-driven calculator that performs different
mathematical operations. The program takes the user’s choice, accepts required numbers,
performs calculations, and displays the result. One operation also uses NumPy arrays.'''
```

```python
'''Logic Explanation:
1. A menu is displayed with different calculator operations.
2. The user selects an operation using input.
3. Based on the choice, the program asks for numbers.
4. For addition, multiple values are stored in a list and converted into a NumPy array.
5. Other operations like subtraction, multiplication, division, power, and modulus are performed normally.
6. Division by zero is handled using a conditional statement.
7. The program runs inside a loop so the user can continue or exit.'''
```

```python
import numpy as np

def show_menu():
    print("\n--- Smart Calculator ---")
    print("1. Addition (uses NumPy)")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")
    print("5. Power")
    print("6. Modulus")
    print("7. Exit")

while True:
    show_menu()
    choice = input("Enter your choice: ").strip()
    if not choice.isdigit():
        print("Please enter a valid number.")
        continue
    choice = int(choice)


    if choice == 1:
        print("\nYou selected Addition: ")
        n = int(input("How many numbers you want to add? "))
        numbers = []

        for i in range(n):
            value = float(input(f"Enter number {i+1}: "))
            numbers.append(value)

        arr = np.array(numbers)
        result = arr.sum()
        print("Result:", result)

    elif choice == 2:
        print("\nYou selected: Subtraction")

        a = float(input("Enter your first number: "))
        b = float(input("Enter your second number: "))

        result = a - b
        print("Result:", result)


    elif choice == 3:
        print("You selected: Multiplication")
        a = float(input("Enter your first number: "))
        b = float(input("Enter your second number: "))
        print("Result:", a * b)

    elif choice == 4:
        print("You selected: Division")
        a = float(input("Enter numerator: "))
        b = float(input("Enter denominator: "))
        if b == 0:
            print("Error: Division by zero not allowed")
        else:
            print("Result:", a / b)

    elif choice == 5:
        print("You selected: Power")
        base = float(input("Enter base: "))
        power = float(input("Enter power: "))
        print("Result:", base ** power)

    elif choice == 6:
        print("You selected: Modulus")
        a = int(input("Enter first number: "))
        b = int(input("Enter second number: "))
        print("Result:", a % b)

    elif choice == 7:
        print("Calculator exited.")
        break

    else:
        print("Invalid choice")
```
